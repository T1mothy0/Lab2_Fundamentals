```python
SURNAME = "Mejos"
SEED_NUM = 5

# Generate and normalize the signal
signal = f"  {SURNAME[0:3]}-{SEED_NUM}  "
processed = signal.strip().upper()
print(f"Generated Signal: '{signal}'")
print(f"Processed Signal: '{processed}'")

letters = 0
digits = 0
spaces = 0
others = 0

for i in range(5):                       # processed signal is 5 characters long
    ch = processed[i]
    code = ord(ch)
    if code >= 65 and code <= 90:        # A-Z
        letters = letters + 1
    elif code >= 48 and code <= 57:      # 0-9
        digits = digits + 1
    elif code == 32:                     # space
        spaces = spaces + 1
    else:
        others = others + 1
    print(f"Character: {ch} | ASCII: {code}")

print(f"Letters: {letters} | Digits: {digits} | Spaces: {spaces} | Others: {others}")

if letters > digits:
    print("Classification: TEXT-DOMINANT")
else:
    print("Classification: NUMERIC-DOMINANT")
```
