```python
SURNAME = "Mejos"
SEED_NUM = 5

valid = 0
invalid = 0
low = 0
nominal = 0
high = 0

for i in range(5):                       # fixed number of readings
    # Generate a reading from the surname and seed
    reading = f"{(ord(SURNAME[i]) * SEED_NUM) % 100}"
    if i == 3:
        reading = "ERR"                  # simulated invalid entry
    print(f"Reading {i + 1}: {reading}")

    # Validate and classify
    try:
        value = int(reading)
        valid = valid + 1
        if value < 20:
            print("VALID - LOW")
            low = low + 1
        elif value > 80:
            print("VALID - HIGH")
            high = high + 1
        else:
            print("VALID - NOMINAL")
            nominal = nominal + 1
    except ValueError:
        print("INVALID - Not numeric")
        invalid = invalid + 1

print(f"Valid: {valid} | Invalid: {invalid}")
print(f"LOW: {low} | NOMINAL: {nominal} | HIGH: {high}")
```
