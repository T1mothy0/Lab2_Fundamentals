```python
SURNAME = "Mejos"
SEED_NUM = 5

password = f"{SURNAME[0:3].upper()}{SEED_NUM * 11}"   # MEJ55
limit = SEED_NUM % 3 + 3                              # 5 attempts
print(f"Generated Password: {password}")
print(f"Attempt Limit: {limit}")

attempts = 0
state = "LOCKED"

while True:
    guess = input("Enter password: ")
    attempts = attempts + 1
    if guess == password:
        state = "UNLOCKED"
        print("Access Granted")
        break
    elif attempts == limit:
        print("Attempt limit reached. Authentication terminated.")
        break
    else:
        print("Access Denied")

print(f"Attempts Made: {attempts}")
print(f"Final System State: {state}")
```
